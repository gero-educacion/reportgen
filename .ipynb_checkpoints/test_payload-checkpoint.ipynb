# reportgen — local payload test harness

Simulates a full `/run` payload going through `process_report_job()` **without** touching real Drive, Sheets, SendGrid, SiteGround, or MySQL. Every external side effect is mocked with a print statement so you can see exactly what *would* have happened, then inspect the actual generated PPTX/PDF on disk.

## Two ways to run this

**A. Inside the container (recommended — full fidelity)**
```bash
docker compose build
docker compose run --rm -v $(pwd):/app -p 8888:8888 reportgen \
    bash -c "pip install jupyter --quiet && jupyter notebook --ip=0.0.0.0 --allow-root --no-browser"
```
Then open the printed URL. `/app/assets`, `/app/tmp`, etc. already exist correctly because you're inside the real image. Set `LOCAL_MODE = False` below.

**B. Bare host (faster iteration, needs one-time setup)**
The code hardcodes `/app/...` paths at import time. To run outside Docker you need those paths to exist and be writable, once:
```bash
sudo mkdir -p /app/tmp /app/data
sudo chown -R $(whoami) /app
```
Then point `LOCAL_ASSETS_DIR` below at wherever your actual `.pptx` templates live on your machine, and set `LOCAL_MODE = True`.

Either way — for **real PDF conversion** (not just a placeholder file), you need a Gotenberg container reachable at `GOTENBERG_URL`:
```bash
docker run -d --rm -p 3000:3000 gotenberg/gotenberg:8
```
That's the whole dependency — no need to build your own image just for this.

## Discover the exact placeholder keys your template expects
Scans the role's `.pptx` for every `<<...>>` token, so you know exactly what to put in the test payload instead of guessing.

````python
import re
from pptx import Presentation

def discover_placeholders(template_path: Path) -> set[str]:
    prs = Presentation(template_path)
    found = set()
    for slide in prs.slides:
        for shape in slide.shapes:
            if not shape.has_text_frame:
                continue
            found.update(re.findall(r"<<(.*?)>>", shape.text_frame.text))
    return found

for suffix, template_path in job_config.get_templates(ROLE_TO_TEST):
    print(f"\n--- {suffix} ({template_path.name}) ---")
    if not template_path.exists():
        print(f"  ⚠️  template not found at {template_path} — check ASSETS_DIR / LOCAL_ASSETS_DIR")
        continue
    for key in sorted(discover_placeholders(template_path)):
        print(f"  <<{key}>>")
````

## Generate a mock payload automatically
Scans every template `ROLE_TO_TEST` uses, collects every `<<PLACEHOLDER>>`, and fills each with a mock value — plus the universal fields every run needs (`student_id`, `job_id`, `Rol`, name, email, and the four flags). For `"full"`/`"utp_chart"` roles, also seeds the raw Holland/Gardner/Áreas de Estudio fields `process_student_data()`/`generate_graph()` expect — those get transformed into derived keys (chart images, presence-level images) before the template ever sees them, so scanning `<<...>>` tokens alone wouldn't catch them.

Re-run this cell anytime you want a fresh random student — pass `seed=` for a reproducible one.

````python
import random

HOLLAND_LABELS = ["REALISTA", "INVESTIGATIVO", "ARTÍSTICO", "SOCIAL", "EMPRENDEDOR", "CONVENCIONAL"]
GARDNER_LABELS = [
    "LINGÜÍSTICO-VERBAL", "LÓGICO-MATEMÁTICA", "VISUAL-ESPACIAL", "CORPORAL-KINESTÉSICA",
    "MUSICAL", "INTERPERSONAL", "INTRAPERSONAL", "NATURALISTA",
]
GARDNER_LEVELS = ["ALTA", "MEDIA", "BAJA"]


def _mock_value_for(key: str) -> str:
    """Best-effort plausible mock value for a single placeholder key.
    No literal <<>> in the output — keeps it distinguishable from a
    genuinely-unfilled placeholder when scanning for leftovers later.
    """
    return f"Acá iría el texto para el placeholder {key}"


def generate_mock_payload(
    rol: str,
    student_id: str = "test-001",
    job_id: str | None = None,
    name: str = "OJ Simpson",
    email: str = "juan@geroeducacion.com",
    seed: int | None = None,
) -> dict:
    if seed is not None:
        random.seed(seed)

    job_id = job_id or f"{student_id}-run1"

    job = {
        "student_id":student_id,
        "job_id": job_id,
        "Rol": rol,
        "Nombre y Apellido": name,
        "Email": email,
        # universal flags — keep every side effect ON so the full flow runs
        "send_email": True,
        "upload_drive": True,
        "post_siteground": True,
        "upload_historic": True,
        "force_rerun": True,
    }

    # 1. scan every template this role renders, collect every <<KEY>>
    placeholder_keys: set[str] = set()
    for suffix, template_path in job_config.get_templates(rol):
        if not template_path.exists():
            print(f"⚠️  template not found, skipping scan: {template_path}")
            continue
        placeholder_keys |= discover_placeholders(template_path)

    # 2. fill each one with a mock value (skip keys already set above)
    for key in sorted(placeholder_keys):
        if key in job:
            continue
        job[key] = _mock_value_for(key)

    # 3. full-pipeline roles ALSO need raw Holland/Gardner/AE fields —
    #    these feed process_student_data()/generate_graph() and don't
    #    appear as literal template tokens (they're transformed first)
    if job_config.needs_full_pipeline(rol):
        raw_pcts = [random.random() for _ in range(6)]
        total = sum(raw_pcts)
        pcts = [p / total for p in raw_pcts]
        for i, (label, pct) in enumerate(zip(HOLLAND_LABELS, pcts), start=1):
            job[f"Holland {i:02d}"] = label
            job[f"% Holland {i:02d}"] = round(pct, 4)
            job[f"PRE Holland {i:02d}"] = random.choice(GARDNER_LEVELS)

        for i, label in enumerate(GARDNER_LABELS, start=1):
            job[f"Gardner {i:02d}"] = label
            job[f"% Gardner {i:02d}"] = random.choice(GARDNER_LEVELS)

        for i in range(1, 9):
            job[f"% AE {i:02d}"] = round(random.uniform(0.1, 0.95), 2)

        for i in range(1, 5):
            job[f"CARRERA_0{i}"] = f"✔️ Carrera de ejemplo #{i}"

    return job


job = generate_mock_payload(ROLE_TO_TEST, seed=42)
print(f"Generated payload with {len(job)} fields:")
job
````

## Batch-test every role
Loops over every role in `ROLE_CONFIGS` (except `_default`, which is a fallback not a real role): checks `role_is_ready()`, and if ready, runs a fully-mocked pipeline with a generated payload. Also scans the output `.pptx` for any leftover `<<...>>` tokens the mock payload didn't cover — a non-zero count there usually means a template placeholder that `discover_placeholders()` found but nothing filled (rare, but possible with nested shapes/tables), OR a role-specific field the generator's Holland/Gardner/AE seeding doesn't fully cover.

One row per role at the end — everything mocked, nothing touches real Drive/Sheets/SiteGround/Email.

````python
import re as _re

def _leftover_tokens(pptx_path):
    prs = Presentation(pptx_path)
    tokens = []
    for slide in prs.slides:
        for shape in slide.shapes:
            if shape.has_text_frame:
                tokens += _re.findall(r"<<.*?>>", shape.text_frame.text)
    return tokens


# make sure everything is mocked before batch-testing every role
tasks_mod.upload_pdf_to_drive        = mock_upload_pdf_to_drive
tasks_mod.upsert_json_to_drive       = mock_upsert_json_to_drive
tasks_mod.send_report_to_siteground  = mock_send_report_to_siteground
tasks_mod.upload_pdf_to_siteground   = mock_noop
tasks_mod.send_report_email          = mock_send_report_email
tasks_mod.update_student_status      = mock_update_student_status
tasks_mod.write_majors_to_db         = mock_noop
tasks_mod.post_utp_payload           = mock_noop
tasks_mod.alter_table_reports        = mock_noop
tasks_mod.upsert_historico           = mock_noop
tasks_mod.download_drive_file        = mock_noop
tasks_mod.get_all_links              = mock_get_all_links
tasks_mod.upsert_student             = mock_noop

ROLES_TO_BATCH_TEST = [r for r in job_config.ROLE_CONFIGS.keys() if r != "_default"]

results = []

for rol in ROLES_TO_BATCH_TEST:
    row = {"rol": rol, "ready": None, "ready_problems": "", "pipeline_ok": None,
           "pipeline_error": "", "leftover": None}

    ready, problems = job_config.role_is_ready(rol)
    row["ready"] = ready
    row["ready_problems"] = "; ".join(problems)

    if not ready:
        results.append(row)
        continue

    try:
        job = generate_mock_payload(rol, student_id=f"batch-{rol}", seed=1)
        result = tasks_mod.process_report_job(job)
        row["pipeline_ok"] = True

        job_dir = tasks_mod.APP_TMP_DIR / job["job_id"]
        leftover_total = 0
        for suffix, _ in job_config.get_templates(rol):
            pptx_path = job_dir / f"report_{suffix}.pptx"
            if pptx_path.exists():
                leftover_total += len(_leftover_tokens(pptx_path))
        row["leftover"] = leftover_total

    except Exception as e:
        row["pipeline_ok"] = False
        row["pipeline_error"] = f"{type(e).__name__}: {e}"

    results.append(row)

print(f"{'ROLE':<18}{'READY':<8}{'PIPELINE':<10}{'LEFTOVER':<10}DETAIL")
print("-" * 90)
for r in results:
    pipeline_str = "—" if r["pipeline_ok"] is None else ("OK" if r["pipeline_ok"] else "FAIL")
    leftover_str = "—" if r["leftover"] is None else str(r["leftover"])
    detail = r["ready_problems"] or r["pipeline_error"]
    print(f"{r['rol']:<18}{str(r['ready']):<8}{pipeline_str:<10}{leftover_str:<10}{detail}")
````
